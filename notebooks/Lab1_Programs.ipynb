```python
from collections import deque
import heapq

# ---------- Grid maze: S = start, G = goal, # = wall ----------
MAZE = ["S . . . . # #",
        "# # . # # . .",
        ". . . # . . #",
        ". # . . . . #",
        ". . . . # . G"]
grid = [row.split() for row in MAZE]
ROWS, COLS = len(grid), len(grid[0])

def find(ch):
    for r in range(ROWS):
        for c in range(COLS):
            if grid[r][c] == ch:
                return (r, c)

def neighbours(cell):
    r, c = cell
    for dr, dc in [(-1, 0), (0, 1), (1, 0), (0, -1)]:   # Up, Right, Down, Left
        nr, nc = r + dr, c + dc
        if 0 <= nr < ROWS and 0 <= nc < COLS and grid[nr][nc] != '#':
            yield (nr, nc)

def bfs(start, goal):
    frontier = deque([start]); parent = {start: None}; expanded = 0
    while frontier:
        node = frontier.popleft()                 # FIFO queue
        expanded += 1
        if node == goal:
            return build_path(parent, goal), expanded
        for nb in neighbours(node):
            if nb not in parent:
                parent[nb] = node
                frontier.append(nb)
    return None, expanded

def dfs(start, goal):
    frontier = [start]; parent = {start: None}; expanded = 0
    visited = set()
    while frontier:
        node = frontier.pop()                     # LIFO stack
        if node in visited:
            continue
        visited.add(node); expanded += 1
        if node == goal:
            return build_path(parent, goal), expanded
        for nb in neighbours(node):
            if nb not in visited:
                parent[nb] = node
                frontier.append(nb)
    return None, expanded

def build_path(parent, node):
    path = []
    while node is not None:
        path.append(node); node = parent[node]
    return path[::-1]

S, G = find('S'), find('G')
print(f"{'Algorithm':<6}{'Path length':>13}{'Nodes expanded':>17}")
for name, algo in [('BFS', bfs), ('DFS', dfs)]:
    path, exp = algo(S, G)
    print(f"{name:<6}{len(path)-1:>13}{exp:>17}")

# ---------- Uniform Cost Search on a weighted graph ----------
graph = {
    'Islamabad':  {'Rawalpindi': 15, 'Murree': 60, 'Attock': 80},
    'Rawalpindi': {'Islamabad': 15, 'Chakwal': 90, 'Jhelum': 110},
    'Murree':     {'Islamabad': 60, 'Abbottabad': 65},
    'Attock':     {'Islamabad': 80, 'Peshawar': 100},
    'Chakwal':    {'Rawalpindi': 90, 'Lahore': 230},
    'Jhelum':     {'Rawalpindi': 110, 'Gujranwala': 110},
    'Gujranwala': {'Jhelum': 110, 'Lahore': 70},
    'Abbottabad': {'Murree': 65},
    'Peshawar':   {'Attock': 100},
    'Lahore':     {'Chakwal': 230, 'Gujranwala': 70},
}

def ucs(graph, start, goal):
    frontier = [(0, start, [start])]          # (path cost, node, path)
    explored = set(); expanded = 0
    while frontier:
        cost, node, path = heapq.heappop(frontier)   # lowest g(n) first
        if node in explored:
            continue
        explored.add(node); expanded += 1
        if node == goal:
            return path, cost, expanded
        for nb, w in graph[node].items():
            if nb not in explored:
                heapq.heappush(frontier, (cost + w, nb, path + [nb]))
    return None, float('inf'), expanded

path, cost, exp = ucs(graph, 'Islamabad', 'Lahore')
print("\nUCS path :", ' -> '.join(path))
print("UCS cost :", cost, "km | nodes expanded:", exp)
```

## Program 3: Arithmetic calculator

Adds, subtracts, multiplies, or divides two numbers.


```python
"""Lab 1 - Program 3: Simple arithmetic calculator."""

print("Simple Calculator")
print("Operations: +  -  *  /")

first = float(input("Enter the first number: "))
operator = input("Enter an operator (+, -, *, /): ").strip()
second = float(input("Enter the second number: "))

if operator == "+":
    result = first + second
elif operator == "-":
    result = first - second
elif operator == "*":
    result = first * second
elif operator == "/":
    if second == 0:
        print("Cannot divide by zero.")
        result = None
    else:
        result = first / second
else:
    print("Unknown operator.")
    result = None

if result is not None:
    print(f"{first} {operator} {second} = {result}")
```

## Program 4: Conditions

Checks even or odd, positive or negative, and assigns a letter grade.


```python
"""Lab 1 - Program 4: Conditions (even/odd and letter grade)."""

number = int(input("Enter an integer: "))

if number % 2 == 0:
    print(f"{number} is even.")
else:
    print(f"{number} is odd.")

if number > 0:
    print("The number is positive.")
elif number < 0:
    print("The number is negative.")
else:
    print("The number is zero.")

marks = float(input("Enter marks out of 100: "))

if marks >= 90:
    grade = "A"
elif marks >= 80:
    grade = "B"
elif marks >= 70:
    grade = "C"
elif marks >= 60:
    grade = "D"
elif marks >= 50:
    grade = "E"
else:
    grade = "F"

print(f"Grade: {grade}")
```

## Program 5: Loops

Prints a multiplication table, a running sum, and a factorial.


```python
"""Lab 1 - Program 5: Loops (multiplication table, sum, and factorial)."""

number = int(input("Enter a positive integer: "))

print(f"\nMultiplication table of {number}")
for i in range(1, 11):
    print(f"{number} x {i} = {number * i}")

total = 0
count = 1
while count <= number:
    total += count
    count += 1
print(f"\nSum of numbers from 1 to {number} = {total}")

factorial = 1
for i in range(1, number + 1):
    factorial *= i
print(f"Factorial of {number} = {factorial}")
```

## Program 6: Lists

```python
"""Lab 1 - Program 6: Lists (store numbers, then summarize them)."""

raw = input("Enter numbers separated by spaces: ")
numbers = [float(item) for item in raw.split()]

if not numbers:
    print("No numbers were entered.")
else:
    print("Numbers:", numbers)
    print("Count:", len(numbers))
    print("Sum:", sum(numbers))
    print("Average:", sum(numbers) / len(numbers))
    print("Minimum:", min(numbers))
    print("Maximum:", max(numbers))
    print("Sorted:", sorted(numbers))
```

## Program 7: Functions

Defines functions for area, prime checking, and temperature conversion.


```python
"""Lab 1 - Program 7: Functions."""


def rectangle_area(length, width):
    return length * width


def is_prime(value):
    if value < 2:
        return False
    for divisor in range(2, int(value ** 0.5) + 1):
        if value % divisor == 0:
            return False
    return True


def celsius_to_fahrenheit(celsius):
    return (celsius * 9 / 5) + 32


length = float(input("Enter rectangle length: "))
width = float(input("Enter rectangle width: "))
print("Area:", rectangle_area(length, width))

candidate = int(input("Enter an integer to test for prime: "))
if is_prime(candidate):
    print(f"{candidate} is prime.")
else:
    print(f"{candidate} is not prime.")

celsius = float(input("Enter temperature in Celsius: "))
print(f"{celsius} C = {celsius_to_fahrenheit(celsius)} F")
```

## Program 8: Rule-based chatbot
