## Program 3: Arithmetic calculator

Adds, subtracts, multiplies, or divides two numbers.


```python
"""Lab 1 - Program 3: Simple arithmetic calculator."""

print("Simple Calculator")
print("Operations: +  -  *  /")

first = float(input("Enter the first number: "))
operator = input("Enter an operator (+, -, *, /): ").strip()
second = float(input("Enter the second number: "))

if operator == "+":
    result = first + second
elif operator == "-":
    result = first - second
elif operator == "*":
    result = first * second
elif operator == "/":
    if second == 0:
        print("Cannot divide by zero.")
        result = None
    else:
        result = first / second
else:
    print("Unknown operator.")
    result = None

if result is not None:
    print(f"{first} {operator} {second} = {result}")
```

## Program 4: Conditions

Checks even or odd, positive or negative, and assigns a letter grade.


```python
"""Lab 1 - Program 4: Conditions (even/odd and letter grade)."""

number = int(input("Enter an integer: "))

if number % 2 == 0:
    print(f"{number} is even.")
else:
    print(f"{number} is odd.")

if number > 0:
    print("The number is positive.")
elif number < 0:
    print("The number is negative.")
else:
    print("The number is zero.")

marks = float(input("Enter marks out of 100: "))

if marks >= 90:
    grade = "A"
elif marks >= 80:
    grade = "B"
elif marks >= 70:
    grade = "C"
elif marks >= 60:
    grade = "D"
elif marks >= 50:
    grade = "E"
else:
    grade = "F"

print(f"Grade: {grade}")
```

## Program 5: Loops

Prints a multiplication table, a running sum, and a factorial.


```python
"""Lab 1 - Program 5: Loops (multiplication table, sum, and factorial)."""

number = int(input("Enter a positive integer: "))

print(f"\nMultiplication table of {number}")
for i in range(1, 11):
    print(f"{number} x {i} = {number * i}")

total = 0
count = 1
while count <= number:
    total += count
    count += 1
print(f"\nSum of numbers from 1 to {number} = {total}")

factorial = 1
for i in range(1, number + 1):
    factorial *= i
print(f"Factorial of {number} = {factorial}")
```

## Program 6: Lists

Reads several numbers into a list and prints count, sum, average, min, and max.


```python
"""Lab 1 - Program 6: Lists (store numbers, then summarize them)."""

raw = input("Enter numbers separated by spaces: ")
numbers = [float(item) for item in raw.split()]

if not numbers:
    print("No numbers were entered.")
else:
    print("Numbers:", numbers)
    print("Count:", len(numbers))
    print("Sum:", sum(numbers))
    print("Average:", sum(numbers) / len(numbers))
    print("Minimum:", min(numbers))
    print("Maximum:", max(numbers))
    print("Sorted:", sorted(numbers))
```

## Program 7: Functions

Defines functions for area, prime checking, and temperature conversion.


```python
"""Lab 1 - Program 7: Functions."""


def rectangle_area(length, width):
    return length * width


def is_prime(value):
    if value < 2:
        return False
    for divisor in range(2, int(value ** 0.5) + 1):
        if value % divisor == 0:
            return False
    return True


def celsius_to_fahrenheit(celsius):
    return (celsius * 9 / 5) + 32


length = float(input("Enter rectangle length: "))
width = float(input("Enter rectangle width: "))
print("Area:", rectangle_area(length, width))

candidate = int(input("Enter an integer to test for prime: "))
if is_prime(candidate):
    print(f"{candidate} is prime.")
else:
    print(f"{candidate} is not prime.")

celsius = float(input("Enter temperature in Celsius: "))
print(f"{celsius} C = {celsius_to_fahrenheit(celsius)} F")
```

## Program 8: Rule-based chatbot

Replies to a few keywords. Type quit to stop.


```python
"""Lab 1 - Program 8: A small rule-based chatbot."""

print("AI Lab Chatbot")
print("Type a message, or type 'quit' to stop.")
print("Try: hello, your name, python, ai, bye")

replies = {
    "hello": "Hello! I am a small rule-based chatbot.",
    "hi": "Hi! How can I help you with Lab 1?",
    "your name": "My name is LabBot. I only know a few replies.",
    "python": "Python is the language we are using in Google Colab.",
    "ai": "Artificial intelligence lets computers learn patterns and make decisions.",
    "bye": "Goodbye! Good luck with your lab.",
}

while True:
    message = input("You: ").strip().lower()
    if message in ("quit", "exit", "bye"):
        print("LabBot: Goodbye!")
        break
    if message in replies:
        print("LabBot:", replies[message])
    else:
        print("LabBot: I do not know that yet. Try hello, python, or ai.")
```
